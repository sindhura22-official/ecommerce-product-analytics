```python
import pandas as pd

file_path = "../data/raw/Online Retail.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Display the column names
print("Columns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Data types:")
print(df.dtypes)
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nMissing-value percentage:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage.round(2))
```

```python
# Check duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
print("Duplicate percentage:", round((duplicate_count / len(df)) * 100, 2), "%")
```

```python
# Check cancelled invoices
cancelled_invoices = df["InvoiceNo"].astype(str).str.startswith("C")

print("Cancelled transaction rows:", cancelled_invoices.sum())
print("Cancellation percentage:", round(cancelled_invoices.mean() * 100, 2), "%")
```

```python
# Check negative quantities
negative_quantity = df["Quantity"] < 0

print("Rows with negative quantity:", negative_quantity.sum())
print(
    "Percentage with negative quantity:",
    round(negative_quantity.mean() * 100, 2),
    "%"
)
```

```python
# Inspect duplicate rows
duplicates = df[df.duplicated(keep=False)]

display(duplicates.head(20))
```

```python
# Inspect cancelled transactions
cancelled = df[df["InvoiceNo"].astype(str).str.startswith("C")]

display(cancelled.head(20))
```

```python
# Negative quantity records
negative_qty = df[df["Quantity"] < 0]

display(negative_qty.head(20))
```

```python
# Compare cancellation and negative-quantity records

print("Cancelled rows:", len(cancelled))
print("Negative quantity rows:", len(negative_qty))

print(
    "\nNegative quantity rows with cancellation invoice:",
    negative_qty["InvoiceNo"].astype(str).str.startswith("C").sum()
)

print(
    "Negative quantity rows without cancellation invoice:",
    (~negative_qty["InvoiceNo"].astype(str).str.startswith("C")).sum()
)
```

```python
# Inspect negative-quantity records that are NOT marked as cancellations

negative_non_cancel = df[
    (df["Quantity"] < 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
]

print("Rows:", len(negative_non_cancel))

display(
    negative_non_cancel[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
    ].head(30)
)
```

```python
# Summarize the non-cancellation negative quantities

print("Unique invoice numbers:",
      negative_non_cancel["InvoiceNo"].nunique())

print("Unique stock codes:",
      negative_non_cancel["StockCode"].nunique())

print("\nQuantity statistics:")
print(negative_non_cancel["Quantity"].describe())

print("\nMost common descriptions:")
display(
    negative_non_cancel["Description"]
    .value_counts(dropna=False)
    .head(10)
)
```

```python
# Check zero and negative unit prices

zero_price = df["UnitPrice"] == 0
negative_price = df["UnitPrice"] < 0

print("Rows with zero UnitPrice:", zero_price.sum())
print("Rows with negative UnitPrice:", negative_price.sum())

print("\nZero-price transactions:")
display(
    df.loc[
        zero_price,
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
    ].head(20)
)

print("\nNegative-price transactions:")
display(
    df.loc[
        negative_price,
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
    ].head(20)
)
```

```python
# Inspect transactions with missing descriptions

missing_description = df[df["Description"].isna()]

print("Rows with missing Description:", len(missing_description))

display(
    missing_description[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
    ].head(20)
)
```

```python
# Inspect negative-price transactions

negative_price_rows = df[df["UnitPrice"] < 0]

print("Negative-price rows:")
display(
    negative_price_rows[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice",
         "CustomerID", "Country"]
    ]
)
```

```python
# Understand zero-price transactions

zero_price_rows = df[df["UnitPrice"] == 0]

print("Zero-price rows:", len(zero_price_rows))

print("\nDescriptions associated with zero-price transactions:")
display(
    zero_price_rows["Description"]
    .value_counts(dropna=False)
    .head(20)
)
```

```python
# Check whether zero-price transactions are cancellations or normal invoices

print(
    "Zero-price cancellation rows:",
    zero_price_rows["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

print(
    "Zero-price positive-quantity rows:",
    (zero_price_rows["Quantity"] > 0).sum()
)

print(
    "Zero-price negative-quantity rows:",
    (zero_price_rows["Quantity"] < 0).sum()
)
```

```python
# Define candidate valid sales

valid_sales_mask = (
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
)

candidate_sales = df[valid_sales_mask].copy()

print("Raw rows:", len(df))
print("Candidate valid sales rows:", len(candidate_sales))
print(
    "Rows excluded:",
    len(df) - len(candidate_sales)
)
print(
    "Percentage retained:",
    round(len(candidate_sales) / len(df) * 100, 2),
    "%"
)
```

```python
print(
    "Candidate sales with missing Description:",
    candidate_sales["Description"].isna().sum()
)

print(
    "Candidate sales with missing CustomerID:",
    candidate_sales["CustomerID"].isna().sum()
)

print(
    "Candidate sales with zero UnitPrice:",
    (candidate_sales["UnitPrice"] == 0).sum()
)

print(
    "Candidate sales with negative Quantity:",
    (candidate_sales["Quantity"] < 0).sum()
)
```

```python
print(
    "Duplicate rows within candidate sales:",
    candidate_sales.duplicated().sum()
)
```

```python
# Inspect duplicate candidate sales

duplicate_sales = candidate_sales[
    candidate_sales.duplicated(keep=False)
].sort_values(
    by=["InvoiceNo", "StockCode", "Description", "Quantity"]
)

display(duplicate_sales.head(30))
```

```python
# Count duplicate groups

duplicate_group_count = (
    candidate_sales
    .groupby(
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"],
        dropna=False
    )
    .size()
)

print("Total duplicate groups:", (duplicate_group_count > 1).sum())

print("\nLargest duplicate group sizes:")
print(
    duplicate_group_count[
        duplicate_group_count > 1
    ].sort_values(ascending=False).head(10)
)
```

```python
# Remove exact duplicate transactions
clean_sales = candidate_sales.drop_duplicates().copy()

print("Rows before duplicate removal:", len(candidate_sales))
print("Rows after duplicate removal:", len(clean_sales))
print("Duplicate rows removed:", len(candidate_sales) - len(clean_sales))
```

```python
# Verify duplicate removal

print("Remaining duplicate rows:", clean_sales.duplicated().sum())
```

```python
# Calculate transaction revenue

clean_sales["Revenue"] = (
    clean_sales["Quantity"] * clean_sales["UnitPrice"]
)

print("Revenue column created.")
display(clean_sales.head())
```

```python
print("Revenue statistics:")
print(clean_sales["Revenue"].describe())

print("\nTotal revenue:")
print(round(clean_sales["Revenue"].sum(), 2))
```

```python
# Save cleaned sales dataset

output_path = "../data/processed/clean_sales.csv"

clean_sales.to_csv(output_path, index=False)

print("Cleaned sales dataset saved successfully!")
print("File:", output_path)
print("Rows:", len(clean_sales))
print("Columns:", len(clean_sales.columns))
```

```python
# Final quality check

print("Shape:", clean_sales.shape)

print("\nMissing values:")
print(clean_sales.isnull().sum())

print("\nDuplicate rows:")
print(clean_sales.duplicated().sum())

print("\nMinimum quantity:")
print(clean_sales["Quantity"].min())

print("\nMinimum unit price:")
print(clean_sales["UnitPrice"].min())

print("\nMinimum revenue:")
print(clean_sales["Revenue"].min())
```

```python
# Create a separate cancellation dataset

cancellations = df[
    df["InvoiceNo"].astype(str).str.startswith("C")
].copy()

cancellations["Revenue"] = (
    cancellations["Quantity"] * cancellations["UnitPrice"]
)

print("Cancellation rows:", len(cancellations))
print("Cancellation revenue/value:", round(cancellations["Revenue"].sum(), 2))
```

```python
# Create special/adjustment transactions dataset

special_transactions = df[
    (df["Quantity"] < 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))
].copy()

print("Special transaction rows:", len(special_transactions))

print("\nTop descriptions:")
display(
    special_transactions["Description"]
    .value_counts(dropna=False)
    .head(15)
)
```

```python
# Save excluded transaction datasets

cancellations.to_csv(
    "../data/processed/cancellations.csv",
    index=False
)

special_transactions.to_csv(
    "../data/processed/special_transactions.csv",
    index=False
)

print("Saved:")
print("- cancellations.csv")
print("- special_transactions.csv")
```
